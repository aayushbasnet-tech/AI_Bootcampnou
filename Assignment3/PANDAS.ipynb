```python
import pandas as bamboo
```

**Series**: a one-dimensional, labeled array-like object capable of holding data of any type

```python
data=[10.2,20.3,30.1,40.5]
series=bamboo.Series(data,index=['a','b','c','d'])
print(series)
print(series.loc['c'])                             #loc=location by label
```

```python
obj=['A','Y','U','S','H']
series=bamboo.Series(obj)
series.loc[4]='A'
print(series)
```

```python
calories={"Day 1":1800,"Day 2":2500,"Day 3":1900}
series=bamboo.Series(calories)
print(series)
series.loc["Day 1"]+=200
print(series[series>=2000])
```

**DATA FRAMES:** a two-dimensional, size-mutable data structure with labeled axes (rows and columns)

```python
data={
    "Name":["Barun","Shishir","Remon"],
      "Age":["60","5","50"]
    }
df=bamboo.DataFrame(data,index=['Employee1','Employee2','Employee3'])
# print(df.loc['Employee1']) OR, using iloc property
print(df.iloc[1])

#add new column
df["Job"]=["Cook","Housekeeper","Washerman"]
print(df)

#add new row
row=bamboo.DataFrame([{"Name":"Yodin","Age":"69","Job":"Sleeper"}],
                     index=['Employee4'])
df=bamboo.concat([df,row])
print(df)


```
