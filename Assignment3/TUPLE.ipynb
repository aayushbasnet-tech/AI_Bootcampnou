**TUPLE**:To store multiple items in a single variable.
          :unchangeable

```python
Royalenfield=("GT650","Hunter350","Meteor","Classic350")
print(Royalenfield)
```

**TUPLE** allow dublicate values

```python
Royalenfield=("GT650","Hunter350","Meteor","Classic350","GT650","Hunter350")
print(Royalenfield)
```

**TUPLE** LENGTH:

```python
Royalenfield=("GT650","Hunter350","Meteor","Classic350")
print(len(Royalenfield))
```

**TUPLE** can be of any data type:

```python
Royalenfield=("GT650","Hunter350","Meteor","Classic350")
print(Royalenfield)
Number=(1, 2, 3, 4, 5)
print(Number)
Bool=(True,False,True)
print(Bool)
```

**TUPLE()** constructor

```python
Royalenfield=tuple(("GT650","Hunter350","Meteor","Classic350"))
print(Royalenfield)
```

Add Items:

```python
SRoyalenfield=("GT650","Hunter350","Meteor","Classic350")
x=list(Royalenfield)
x.append("Himalayan")
Royalenfield=tuple(x)
print(Royalenfield)
```

Remove items:

```python
Royalenfield=("GT650","Hunter350","Meteor","Classic350")
x=list(Royalenfield)
x.remove("Meteor")
Royalenfield=tuple(x)
print(Royalenfield)
```

**SOME EXAMPLES OF TUPLES**

CHECK "**Meteor**" exists in a tuple

```python
Royalenfield=("GT650","Hunter350","Meteor","Classic350")
if "Meteor" in Royalenfield:
    print("Meteor found")
else:
    print("Meteor Not Found")
```

**REVERSING** the TUPLE

```python
numbers = (10, 20, 30, 40, 50)
print("Original Tuple:", numbers)
print("Reversed Tuple:", numbers[::-1])
```

**COUNTING** the number appearence

```python
numbers = (10,20,10,30,10,40)
count = numbers.count(10)
print("10 appears",count,"times.")
```
