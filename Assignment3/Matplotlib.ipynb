```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

**Simple line plot:**

```python
x=[1, 2, 3, 4, 5]
y=[10, 15, 13, 20, 18]
plt.plot(x, y)
plt.show()
```

**Multiple Lines:**

```python
x=[1,2,3,4,5]
y1=[2,4,6,8,10]
y2=[1,3,5,7,9]
plt.plot(x, y1)
plt.plot(x, y2)
plt.show()
```

**PLOT CUSTOMIZATION:**

```python
plt.plot([1,2,3,4],[4,8,2,6],
    color="green",
    linestyle="--",
    marker="o",
    linewidth=3,
    markersize=10)
plt.show()
```

Labels

```python
x=[1,2,3,4]
y=[5,9,7,12]
plt.plot(x,y)
plt.title("Sales")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

Legend

```python
x=[1,2,3,4]
plt.plot(x,[2,4,6,8],label="Product A")
plt.plot(x,[1,3,5,7],label="Product B")
plt.legend()
plt.show()
```

Grid lines

```python
plt.plot([1,2,3,4],[4,8,2,6])
plt.grid(color="gray",linestyle="--",linewidth=0.5)
plt.show()
```

**BARCHARTS:**

```python
students=["Ram","Hari","Sita","Gita"]
marks=[80,65,90,70]
plt.bar(students,marks)
plt.show()
```

Horizontal BarCharts with colored bars

```python
colors=["red","green","blue","orange"]
plt.barh(students,marks,color=colors)
plt.show()
```

**PieCharts:**

```python
subjects=["Math","Science","English"]
marks=[40,35,25]
plt.pie(marks,labels=subjects,autopct="%1.1f%%")
plt.show()
```

Explode Slice

```python
explode=[0.2,0,0]
plt.pie(marks,labels=subjects,autopct="%1.1f%%",explode=explode)
plt.show()
```

**SCATTER PLOT:**

```python
x=[1,2,3,4,5]
y=[5,7,4,9,10]
plt.scatter(x,y,color="red",s=150)
plt.show()
```

**HISTORGRAM:**

```python
data=np.random.randn(1000)
plt.hist(data,bins=20,color="orange",edgecolor="black")
plt.show()
```

**SUBPLOTS:**

```python
plt.subplot(1,2,1)
plt.plot([1,2,3],[2,4,6])
plt.title("Plot 1")
plt.subplot(1,2,2)
plt.plot([1,2,3],[3,2,5])
plt.title("Plot 2")
plt.show()
```

2*2 GRID

```python
plt.subplot(2,2,1)
plt.plot([1,2,3],[1,2,3])
plt.subplot(2,2,2)
plt.bar(["A","B"],[4,7])
plt.subplot(2,2,3)
plt.scatter([1,2,3],[3,2,5])
plt.subplot(2,2,4)
plt.pie([30,40,30])
plt.tight_layout()
plt.show()
```

**MATPLOTLIB + PANDAS:**

```python
df = pd.DataFrame({
    "Month": ["Jan","Feb","Mar","Apr","May"],
    "Sales": [100,120,90,150,180],
    "Profit": [20,35,18,40,50]
})
plt.figure(figsize=(8,5))
plt.plot(df["Month"], df["Sales"],marker="o",linewidth=2,color="blue",label="Sales")
plt.plot(df["Month"], df["Profit"],marker="s",linewidth=2,color="red",label="Profit")
plt.title("Monthly Report")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.grid(True)
plt.legend()
plt.show()
```
