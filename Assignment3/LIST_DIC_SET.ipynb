**SET**: UNCHANGABLE

```python
set={"Football", "Circket", "Basketball"}
print(set)
```

Access **Set** Items:

```python
set={"Football", "Circket", "Basketball"}
print("Football" in set)
print("Circket" not in set)
```

Add **SET** items

```python
set={"Football", "Circket", "Basketball"}
set.add("Badminton")
seet={"Chess", "LUDO", "Carrom"}
set.update(seet)
print(set)
```

Remove **SET** items

```python
set={"Football", "Circket", "Basketball"}
set.remove("Basketball")
print(set)
```

Join **SETS**

```python
set1={"a", "b", "c"}
set2={1, 2, 3}

set3=set1.union(set2)
print(set3)
```
